```python
import tensorflow as tf
from tensorflow import keras

# Load the SavedModel (as a computation graph)
saved_model_path = "weights/audio_model"
loaded = tf.saved_model.load(saved_model_path)

# Check available signatures
print("Available signatures:", loaded.signatures.keys())

# Get the inference function
infer = loaded.signatures["serving_default"]

# Get input and output shapes
input_shape = list(infer.structured_input_signature[1].values())[0].shape
output_shape = list(infer.structured_outputs.values())[0].shape
dtype = list(infer.structured_input_signature[1].values())[0].dtype

# Wrap the model in a Keras Functional API model
inputs = keras.Input(shape=input_shape[1:], dtype=dtype)
outputs = infer(inputs)
model = keras.Model(inputs=inputs, outputs=outputs)

# Save in .keras format
model.save("weights/audio_model.keras")
print("Model converted and saved as weights/audio_model.keras")
```
